## Step 2: Load the Dataset

The Titanic dataset is loaded into a Pandas DataFrame. This dataset contains information about passengers such as age, gender, passenger class, fare, and survival status. Loading the dataset is the first step before performing any analysis.

```python
df = pd.read_csv("tested.csv")
```

## Step 3: Explore the Dataset

Before cleaning the data, it is important to understand its structure.

The following functions are used:

- **head()** displays the first five rows.
- **info()** provides information about the dataset including data types and missing values.
- **shape** returns the total number of rows and columns.

### Observation
The dataset contains 418 rows and 12 columns. It includes both numerical and categorical variables. Some columns contain missing values that need to be handled before analysis.

```python
df.head()
df.info()
df.shape
```

## Step 4: Check for Missing Values

Missing values can affect the accuracy of data analysis and machine learning models. This step identifies the number of missing values present in each column.

```python
df.isnull().sum()
```

### Observation

The dataset contains missing values in the following columns:

- Age
- Fare
- Cabin

The Cabin column has a large number of missing values, while Age and Fare have only a few missing entries.

## Step 5: Check for Duplicate Records

Duplicate records can lead to inaccurate analysis and biased results. This step checks whether duplicate rows exist in the dataset.

```python
df.duplicated().sum()
```

### Observation

No duplicate rows were found in the dataset. Therefore, no records need to be removed.

## Step 6: Handle Missing Values

The missing values are handled using appropriate techniques.

- The **Age** column is filled using the median because it is less affected by extreme values.
- The **Fare** column is also filled using the median since only one value is missing.
- The **Cabin** column is dropped because most of its values are missing, making it unsuitable for meaningful analysis.

```python
df['Age'].fillna(df['Age'].median(), inplace=True)
```

Fill Fare

```python
df['Fare'].fillna(df['Fare'].median(), inplace=True)
```

Drop Cabin

```python
df.drop('Cabin', axis=1, inplace=True)
```

## Step 7: Verify Data Types

The data types of all columns are checked to ensure that numerical and categorical variables are stored correctly for further analysis.

```python
df.dtypes
```

### Observation

All columns have appropriate data types. Therefore, no data type conversion was required.

## Step 8: Generate Summary Statistics

Summary statistics provide a quick overview of the dataset.

This analysis includes:

- Mean
- Median
- Standard Deviation
- Minimum and Maximum Values

Additionally, frequency counts are generated for categorical variables such as gender and passenger class.

```python
df.describe()
df['Sex'].value_counts()
df['Pclass'].value_counts()
```

### Observation

The descriptive statistics provide useful insights into passenger ages, fares, and class distribution. Frequency counts also show the number of passengers in each category.

## Step 9: Histogram of Passenger Ages

A histogram is created to visualize the distribution of passenger ages. This helps identify the most common age groups and understand the spread of the data.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
plt.hist(df['Age'], bins=20)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')
plt.show()
```

### Observation

Most passengers are young to middle-aged adults. The age distribution is not perfectly symmetrical and shows that fewer passengers belong to the older age groups.

## Step 10: Passenger Class Distribution

A bar chart is created to display the number of passengers in each passenger class. This visualization makes it easy to compare the class distribution.

```python
import seaborn as sns

plt.figure(figsize=(6,4))
sns.countplot(x='Pclass', data=df)
plt.title('Passengers by Class')
plt.xlabel('Passenger Class')
plt.ylabel('Count')
plt.show()
```

### Observation

The chart shows the distribution of passengers across different travel classes. Some passenger classes contain more travelers than others, indicating an unequal distribution.
